```python
from pyspark.sql import functions as F
from pyspark.sql.types import StructType,StructField,StringType,DoubleType
```

```python
circuits_schema = StructType([
    StructField("circuitId",StringType()),
    StructField("url",StringType()),
    StructField("circuitName",StringType()),
    StructField("lat",DoubleType()),
    StructField("long",DoubleType()),
    StructField("locality",StringType()),
    StructField("country",StringType())
])
```

```python
circuits_df = spark.read \
    .format("csv") \
    .option("mode", "FAILFAST") \
    .option("header","true") \
    .schema(circuits_schema) \
    .load("/Volumes/formula1/landing/files/circuits.csv")
```

```python
circuits_df=circuits_df.withColumn('ingestion_timestamp',F.current_timestamp()) \
    .withColumn('source_file',F.col("_metadata.file_path"))
```

```python
# %sql
# Drop TABLE if exists formula1.bronze.circuits
```

```python
circuits_df.write.mode("overwrite").saveAsTable('formula1.bronze.circuits')
```
