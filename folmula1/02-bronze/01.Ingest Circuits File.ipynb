```python
circuits_schema = StructType([
    StructField("circuitId",StringType()),
    StructField("url",StringType()),
    StructField("circuitName",StringType()),
    StructField("lat",DoubleType()),
    StructField("long",DoubleType()),
    StructField("locality",StringType()),
    StructField("country",StringType())
])
```

```python
circuit_file_path = f"{landing_folder_path}/circuits.csv"

circuits_df = spark.read \
    .format("csv") \
    .option("mode", "FAILFAST") \
    .option("header","true") \
    .schema(circuits_schema) \
    .load(circuit_file_path)
```

```python
circuits_df = add_ingestion_metadata(circuits_df)
```

```python
circuits_table = f'{catalog_name}.{bronze_schema}.circuits'
circuits_df.write.mode("overwrite").saveAsTable(circuits_table)
```
