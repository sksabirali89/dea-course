```python
from pyspark.sql import functions as F
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,DateType
```

```python
constructors_schema = """
constructorId STRING,
name STRING,
nationality STRING,
url STRING
"""
```

```python
constructors_df = spark.read \
    .format("json") \
    .option("mode", "FAILFAST") \
    .option("header","true") \
    .schema(constructors_schema) \
    .load("/Volumes/formula1/landing/files/constructors.json")
```

```python
constructors_df=constructors_df.withColumn('ingestion_timestamp',F.current_timestamp()) \
    .withColumn('source_file',F.col("_metadata.file_path"))
```

```python
# %sql
# Drop TABLE if exists formula1.bronze.constructors
```

```python
constructors_df.write.mode("overwrite").saveAsTable('formula1.bronze.constructors')
```
