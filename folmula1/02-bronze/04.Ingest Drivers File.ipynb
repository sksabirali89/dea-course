```python
from pyspark.sql import functions as F
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,DateType
```

```python
name_schema = StructType([
    StructField("familyName",StringType()),
    StructField("givenName",StringType())
])

drivers_schema = StructType([
    StructField("driverId",StringType()),
    StructField("name",name_schema),
    StructField("dateOfBirth",DateType()),
    StructField("nationality",StringType()),
    StructField("url",StringType())
])
```

```python
drivers_df = spark.read \
    .format("json") \
    .option("mode", "FAILFAST") \
    .option("header","true") \
    .schema(drivers_schema) \
    .load("/Volumes/formula1/landing/files/drivers.json")
```

```python
drivers_df=drivers_df.withColumn('ingestion_timestamp',F.current_timestamp()) \
    .withColumn('source_file',F.col("_metadata.file_path"))
```

```python
# %sql
# Drop TABLE if exists formula1.bronze.drivers
```

```python
drivers_df.write.mode("overwrite").saveAsTable('formula1.bronze.drivers')
```
