```python
from pyspark.sql import functions as F
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,DateType
```

```python
races_schema = StructType([
    StructField("season",IntegerType()),
    StructField("round",IntegerType()),
    StructField("url",StringType()),
    StructField("raceName",StringType()),
    StructField("date",DateType()),
    StructField("circuitId",StringType())
])
```

```python
races_df = spark.read \
    .format("csv") \
    .option("mode", "FAILFAST") \
    .option("header","true") \
    .schema(races_schema) \
    .load("/Volumes/formula1/landing/files/races.csv")
```

```python
races_df=races_df.withColumn('ingestion_timestamp',F.current_timestamp()) \
    .withColumn('source_file',F.col("_metadata.file_path"))
```

```python
%sql
Drop TABLE if exists formula1.bronze.races
```

```python
races_df.write.mode("overwrite").saveAsTable('formula1.bronze.races')
```
