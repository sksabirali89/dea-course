```python
from pyspark.sql import functions as F
```

```python
circuits_df = spark.table('formula1.bronze.circuits')
```

```python
circuits_df.printSchema()
```

Drop URL from circuits_df dataframe

```python
circuits_df = circuits_df.select(
    F.col("circuitId").alias("circuit_id"),
    F.col("circuitName").alias("circuit_name"),
    F.col("lat").alias("latitude"),
    F.col("long").alias("longitude"),
    F.col("locality"),F.col("country"),
    F.col("ingestion_timestamp"),
    F.col("source_file")
    )
```

```python
circuits_df = circuits_df.dropDuplicates()
```

```python
circuits_df = circuits_df.filter(F.col("circuitId").isNotNull())
```

```python
circuits_df = circuits_df.withColumn(
    "circuit_name", F.initcap(F.col("circuit_name"))
).withColumn("locality",F.initcap(F.col("locality")))
```

```python
circuits_df.write.mode("overwrite").saveAsTable('formula1.silver.circuits')
```
