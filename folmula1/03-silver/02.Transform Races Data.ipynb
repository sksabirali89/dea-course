```python
races_df = spark.table('formula1.bronze.races')
```

```python
races_df.printSchema()
```

Drop URL from races_df dataframe

```python
races_df = races_df.select(
    F.col("season"),
    F.col("round"),
    F.col("raceName").alias("race_name"),
    F.col("date").alias("race_date"),
    F.col("circuitId").alias("circuit_id"),
    F.col("ingestion_timestamp"),
    F.col("source_file")
    )
```

```python
races_df = races_df.dropDuplicates()
```

```python
races_df = races_df.withColumn(
    "race_name", F.initcap(F.col("race_name"))
)
```

```python
target_table = f'{catalog_name}.{silver_schema}.races'
races_df.write.mode("overwrite").saveAsTable(target_table)
```
