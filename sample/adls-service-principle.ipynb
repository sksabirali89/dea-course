### Accessing Azure Data Lake Storage Gen2 from Databricks

This notebook demonstrates how to configure Spark to access Azure Data Lake Storage Gen2 using OAuth authentication. The configuration uses a service principal and client secret to authenticate.

> **Note:**  
> This approach does **not** work with Databricks serverless clusters.  
> For serverless or production workloads, it is recommended to use [Unity Catalog](https://docs.databricks.com/en/data-governance/unity-catalog/index.html) with an [external location](https://docs.databricks.com/en/data-governance/unity-catalog/manage-external-locations.html) for secure and scalable data access.

**Steps covered:**
1. Set up Spark configuration for OAuth authentication.
2. Read data from Azure Data Lake Storage Gen2 using the `abfss` protocol.

```python
# service_credential = dbutils.secrets.get(scope="<scope>",key="<service-credential-key>")
service_credential=''
 
spark.conf.set("fs.azure.account.auth.type.deacoursedl1.dfs.core.windows.net", "OAuth")
spark.conf.set("fs.azure.account.oauth.provider.type.deacoursedl1.dfs.core.windows.net", "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider")
spark.conf.set("fs.azure.account.oauth2.client.id.deacoursedl1.dfs.core.windows.net", "81aec2f5-7124-41d6-8ea1-aece3fc8f2ce")
spark.conf.set("fs.azure.account.oauth2.client.secret.deacoursedl1.dfs.core.windows.net", service_credential)
spark.conf.set("fs.azure.account.oauth2.client.endpoint.deacoursedl1.dfs.core.windows.net", "https://login.microsoftonline.com/cf581206-3d23-48bd-8d73-a12932a607f0/oauth2/token")
```

```python
# abfss is the Azure Blob File System Secure protocol used to access Azure Data Lake Storage Gen2
df = spark.read.csv("abfss://demo@deacoursedl1.dfs.core.windows.net/circuit.csv")
```
